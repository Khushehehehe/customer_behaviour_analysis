```python
median_review_rating_per_category = df.groupby('Category')['Review Rating'].median()

def fill_review_rating(row):
    if pd.isnull(row['Review Rating']):
        return median_review_rating_per_category[row['Category']]
    return row['Review Rating']

df['Review Rating'] = df.apply(fill_review_rating, axis=1)
```

After filling the null values, let's verify if there are any remaining null values in the 'Review Rating' column.

```python
display(df.isnull().sum())
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
#create a column age_group
labels = ['Young Adult','Adult','Middle Aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4, labels= labels)
df[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Bi-Monthly': 15,
    'Quaterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['frequency_of_purchases','purchase_frequency_days']].head(10)
```

```python
#checking if discounts applied and promo code used are same or not
(df['discount_applied'] == df['promo_code_used']).value_counts()
```

```python
df = df.drop('promo_code_used',axis=1)
```
